# Day 1 — RAG, embeddings, and semantic search


## Goal
Understand how a RAG system converts text into vectors and retrieves passages by meaning.

## Study and understand the following terms:

1. What RAG is.
   - A RAG system does not teach an LLM new information. It simply finds the right information first, then gives it to the LLM before it answers.
   - Think of RAG as an open-book exam instead of a closed-book exam.
   - RAG stands for Retrieval-Augmented Generation. It has two main jobs:
   - Retrieve the most relevant information.
   - Generate an answer using that information.
2. Documents, chunks, embeddings, and queries
   - These are the building blocks of every RAG system.
   - Documents is simply the original information (PDFs, Word docs, emails, medical records, web pages, user manuals, text files, etc.)
   - Chunks. LLMs don't usually search entire documents. Instead, documents are broken into smaller pieces called chunks. Searching small chunks is faster, cheaper and more accurate.
   - Embeddings. Computers don't understand meaning, so we convert every chunk into numbers. These numbers are called an embedding. Similar meanings produce similar vectors.
   - Queries. A query is simply the user's question. Questions are also converted into an embedding, so the system can compare the query to every document chunk.
4. Semantic search
   - Semantic search looks for matching meaning whereas traditional search looks for matching words.
5. Vector similarity
   - Every chunk becomes a point in a huge mathematical space. Similar topics are close together. Different topics are far apart.
   - The system measures: "How close are these two vectors?". Closer vectors = higher similarity = better search result.
6. Cosine similarity
   - Cosine similarity is the most common way to compare vectors.
   - Cosine similarity only asks "Are these vectors pointing in the same direction?"; if yes, they probably have similar meaning. If not, not related.
7. Bi-Encoders
   - Is an embedding model that encodes the question and the document separately: question -> encoder -> vector. document -> encoder -> vector.
   - Popular bi-encoders: Sentence Transformers, BGE, E5, MiniLM
8. Why retrieval is separate from generation.
   - Retrieval: find information. Generation: explain the information.
   - Keeping them separate has many benefits:
      - better accuracy: reads evidence and answers whereas only LLM's memory may hallucinate.
      - easy updates: if documents change; without RAG you need to retrain/fine-tune model; with RAG simply update the documents, re-index it, done.
      - better citations: LLM receives the original document it can answer directly instead of guessing. 
      - more trustworthy: if retrieval finds nothing useful, the system can respond "I couldn't find enough information to answer that" instead of inventing an answer.

### Compatibility rule

Use the same compatible embedding setup for both indexing and querying. Changing the embedding model usually requires rebuilding the vector index.

```
$ export RUN_MODEL_TESTS=1
$ pytest tests/unit/test_embeddings.py -v```

### Required experiments

Run at least five queries and record the top passage, score, and whether it actually contains useful evidence. Add one misleading but semantically related passage and inspect where it ranks.

1. What health condition does the patient have?
    1. score=0.5225 | The patient has a history of hypertension.

2. Does the patient have blood pressure issues?
    1. score=0.7369 | The patient has a history of hypertension.
       
3. Does the patient have kidney disease?
    1. The patient has a history of hypertension
       
4.  Does the patient have sleep apnea?
    1. score=0.4291 | The patient has a history of hypertension.
  
5. Is lisinopril an effective drug for hypertension?
    1. score=0.8176 | Lisinopril is known to not cure hypertension.

### Reflection questions

#### Why are vectors useful for semantic search?
A: Because they convert complex, unstructured data (text or images) into numerical representations of meaning. Search systems calculate mathematical distances between vectors, allowing them to find results that share conceptual context, synonyms or related themes.

#### Why may blood pressure medicine match lisinopril?
A: Because the model may have correlated the word lisinopril with blood pressure medicine from contextual co-occurrence across medical literature, notes or textbooks. They have semantic equivalency (in the sentence 'The patient was prescribed [prescription] to lower their hypertension', replacing 'prescription' with either 'lisinopril' or 'blood pressure medicine' maintains perfect semantic and grammatical validity). 

#### Why must query and passage embeddings be compatible?
A: Because the search and retrieval process fundamentally relies on measuring the geometric distance (cosine similarity) between vectors. If the embeddings are generated using different models or configurations, their numerical values exist in completely different coordinate systems, making comparison mathematically meaningless.

#### Why is similarity not answer confidence?
A: Because high similarity does not guarantee truth; a document can be highly similar to a query but contain incorrect information, fabricated claims, or plagiarized text.

#### Why test retrieval before adding an LLM?
A: Because it ensures the Retrieval-Augmented Generation (RAG) system is built on accurate data rather than flawed assumptions. If your retrieval system fetches the wrong information, even the most advanced LLM will generate incorrect, hallucinated, or irrelevant answers.

### Quick Summary
- RAG:	Finds the right information first, then lets the LLM answer using that information.
- Document:	The original source (PDF, webpage, email, medical record, etc.).
- Chunk:	A small section of a document that is easier to search.
- Embedding:	A list of numbers that represents the meaning of text.
- Query:	The user's question, also converted into an embedding.
- Semantic: Search	Searches by meaning rather than exact words.
- Vector Similarity:	Measures how close two embeddings are in meaning.
- Cosine Similarity:	A common mathematical method for measuring how similarly two vectors point, which approximates semantic similarity.
- Bi-Encoder:	Encodes queries and documents separately so millions of document embeddings can be precomputed and searched quickly.
- Retrieval vs. Generation:	Retrieval finds relevant evidence; generation uses that evidence to produce a natural-language answer. Keeping them separate makes the system more accurate, easier to update, and less prone to hallucinations.
